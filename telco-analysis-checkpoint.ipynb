```python
fig6 = px.histogram(df, x='tenure', title='Distribution of Customer Tenure')
fig6.show()
```

```python
'''
Customer tenure follows a "bathtub" distribution — a large spike of 
customers near 0 months (new signups) and another rise near 70+ months 
(long-term loyal customers), with a dip in the middle. This suggests 
churn risk is highest in the early-to-mid period of the customer 
lifecycle, not evenly spread across all tenures.
'''
```

```python
fig7 = px.histogram(df, x='MonthlyCharges', title='Distribution of Monthly Charges')
fig7.show()
```

```python
'''
Monthly charges show two clusters
— a spike near $20 (customers on minimal/basic plans) and 
a broader hump around $70-$100 (customers with multiple bundled services, 
likely Fiber optic + add-ons).
This aligns with the earlier finding that Fiber optic customers 
churn at a much higher rate, suggesting price sensitivity may be a 
contributing factor.
'''
```

```python
fig8 = px.box(df, x='Churn', y='MonthlyCharges', title='Monthly Charges by Churn Status')
fig8.show()
```

```python
'''
Churned customers have a higher median MonthlyCharges (~$78) than 
retained customers (~$63), suggesting price sensitivity contributes 
to churn — consistent with the earlier finding that pricier Fiber optic
customers churn more.
'''
```

```python
df['Churn_numeric'] = df['Churn'].replace({'No': 0, 'Yes': 1})
```

```python
corr_matrix = df[['tenure', 'MonthlyCharges', 'TotalCharges', 'SeniorCitizen', 'Churn_numeric']].corr()
print(corr_matrix)
```

```python
import plotly.express as px
fig9 = px.imshow(corr_matrix, text_auto=True, color_continuous_scale='RdBu_r',
                  title='Correlation Heatmap')
fig9.show()
```

```python
'''
Tenure has the strongest correlation with churn (-0.35) 
— longer-tenured customers are significantly less likely to churn, 
consistent with the "bathtub" pattern seen in the tenure histogram.
MonthlyCharges shows a weaker positive correlation (+0.19),
supporting the earlier finding that churned customers tend to pay more.
TotalCharges' negative correlation (-0.20) is largely explained by its
strong relationship with tenure (0.83) rather than an independent effect.
'''
```
